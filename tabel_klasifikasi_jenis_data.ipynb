```python
import os, json, wave, struct, math
import pandas as pd
from PIL import Image
pd.set_option("display.max_colwidth", None)
```

## 1. Tabel Klasifikasi

```python
klasifikasi = pd.DataFrame([
    # --- Berkas pada soal ---
    ["destinasi.csv",  "Structured",                     "pd.read_csv",                                   "Basis data relasional"],
    ["destinasi.xlsx", "Structured",                     "pd.read_excel (butuh openpyxl)",                "Basis data relasional (impor ke tabel)"],
    ["ulasan.json",    "Semi-structured",                "json.load / pd.read_json",                      "Basis data dokumen (NoSQL, mis. MongoDB) atau kolom JSONB di PostgreSQL"],
    ["katalog.xml",    "Semi-structured",                "xml.etree.ElementTree / pd.read_xml",           "Basis data dokumen / basis data XML"],
    ["contoh.png",     "Unstructured (gambar)",          "PIL.Image.open / matplotlib.pyplot.imread",     "Object storage / file system; metadata di basis data"],
    ["teks bebas",     "Unstructured (teks)",            "open(...).read() / pathlib.Path.read_text",     "Object storage / data lake; untuk pencarian: mesin pencari teks (mis. Elasticsearch)"],
    # --- Tambahan (cari sendiri, topik warung Denpasar) ---
    ["warung.geojson",     "Semi-structured (spasial)",  "json.load / geopandas.read_file",               "Basis data spasial (PostgreSQL + PostGIS)"],
    ["foto_makanan.jpg",   "Unstructured (gambar)",      "PIL.Image.open",                                "Object storage; metadata dan URL di basis data"],
    ["ulasan_suara.wav",   "Unstructured (audio)",       "wave.open / librosa.load",                      "Object storage; transkrip teks di basis data"],
    ["menu_warung.pdf",    "Unstructured (dokumen)",     "pdfplumber.open / PyPDF2.PdfReader",            "Object storage / data lake"],
], columns=["Berkas", "Jenis Data", "Cara Membaca", "Storage yang Sesuai"])
klasifikasi.set_index("Berkas")
```

```python
os.makedirs("docs", exist_ok=True)
klasifikasi.to_csv("docs/klasifikasi_jenis_data.csv", index=False, encoding="utf-8")
print("Tersimpan:", os.path.abspath("docs/klasifikasi_jenis_data.csv"))
pd.read_csv("docs/klasifikasi_jenis_data.csv").head()
```

```python
os.makedirs("sampel_data", exist_ok=True)

# Berkas contoh
json.dump([{"warung": "Warung Nasi Ayam Kedewatan", "rating": 5}],
          open("sampel_data/ulasan.json", "w", encoding="utf-8"), ensure_ascii=False)
json.dump({"type": "FeatureCollection", "features": [
    {"type": "Feature", "properties": {"nama": "Warung Sanur"},
     "geometry": {"type": "Point", "coordinates": [115.2630, -8.6780]}}]},
    open("sampel_data/warung.geojson", "w"))
Image.new("RGB", (640, 480), (230, 160, 60)).save("sampel_data/foto_makanan.jpg")
sr = 22050
with wave.open("sampel_data/ulasan_suara.wav", "wb") as w:
    w.setnchannels(1); w.setsampwidth(2); w.setframerate(sr)
    w.writeframes(b"".join(struct.pack("<h", int(12000*math.sin(2*math.pi*440*t/sr))) for t in range(sr*2)))
pd.DataFrame({"destinasi": ["Tanah Lot", "Sanur"], "kota": ["Tabanan", "Denpasar"]}).to_csv("sampel_data/destinasi.csv", index=False)

# Cara membaca
print("pd.read_csv ->", pd.read_csv("sampel_data/destinasi.csv").shape)
print("json.load   ->", json.load(open("sampel_data/ulasan.json", encoding="utf-8"))[0]["warung"])
print("geojson     ->", len(json.load(open("sampel_data/warung.geojson"))["features"]), "fitur")
with Image.open("sampel_data/foto_makanan.jpg") as im:
    print("PIL.open    ->", im.size, im.format)
with wave.open("sampel_data/ulasan_suara.wav") as w:
    print("wave.open   ->", w.getframerate(), "Hz,", w.getnframes()/w.getframerate(), "detik")
```
